````python
from typing import Any, Dict, List
from typing_extensions import TypedDict
from datetime import datetime
import json

from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage
from langgraph.graph import StateGraph, END

print("✅ ResearchFlow imports successful")


class ResearchState(TypedDict, total=False):
    query: str
    research_plan: List[str]
    search_results: Dict[str, Any]
    retrieved_docs: List[Dict[str, Any]]
    reasoning: str
    final_answer: str
    citations: List[Dict[str, str]]
    memory: Dict[str, Any]


llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0.3,
    api_key=os.environ["GROQ_API_KEY"]
)

print("✅ ResearchFlow state initialized")
print("✅ Groq LLM initialized")
````

````python
def query_planner_agent(state: ResearchState) -> ResearchState:
    """
    Break the user's research query into focused sub-questions.
    """

    query = state["query"]

    prompt = f"""
You are the Query Planning Agent in a research assistant system.

User research query:
{query}

Break this query into 3 to 4 specific research sub-questions.

Requirements:
- Each sub-question must investigate a different aspect of the main query.
- Make the questions specific and researchable.
- Avoid duplicate questions.
- Return ONLY valid JSON.
- Do not use Markdown or code fences.

Required format:
{{
    "sub_questions": [
        "Question 1",
        "Question 2",
        "Question 3",
        "Question 4"
    ]
}}
"""

    response = llm.invoke([HumanMessage(content=prompt)])
    content = response.content.strip()

    try:
        # Handle accidental Markdown code fences
        if content.startswith("```"):
            content = content.replace("```json", "").replace("```", "").strip()

        plan_obj = json.loads(content)

        sub_questions = plan_obj.get("sub_questions", [])

        if not isinstance(sub_questions, list) or not sub_questions:
            raise ValueError("Invalid sub_questions format")

        state["research_plan"] = sub_questions[:4]

    except Exception as e:
        print(f"⚠️ Planner JSON parsing issue: {e}")
        print("Raw response:", content)

        # Safe fallback
        state["research_plan"] = [query]

    return state


print("✅ Query Planner Agent created")
````

````python
test_state: ResearchState = {
    "query": "What are the latest methods for detecting diabetic retinopathy using computer vision?"
}

test_state = query_planner_agent(test_state)

print("🔎 Original Query:")
print(test_state["query"])

print("\n📋 Generated Research Plan:")

for i, question in enumerate(test_state["research_plan"], 1):
    print(f"{i}. {question}")

print("\n✅ Query Planner test completed")
````

````python
from langchain_community.utilities import DuckDuckGoSearchAPIWrapper

search = DuckDuckGoSearchAPIWrapper()

def retrieval_agent(state: ResearchState) -> ResearchState:
    """
    Search the web for each research sub-question
    and collect the retrieved documents.
    """

    retrieved_docs = []

    research_plan = state.get("research_plan", [])

    for sub_question in research_plan[:4]:

        print(f"\n🔍 Searching: {sub_question}")

        try:
            results = search.results(
                sub_question,
                num_results=3
            )

            for result in results:
                document = {
                    "title": result.get("title", "Unknown"),
                    "snippet": result.get("snippet", ""),
                    "url": result.get("link", ""),
                    "query": sub_question,
                    "retrieved_at": datetime.now().isoformat()
                }

                retrieved_docs.append(document)

        except Exception as e:
            print(f"⚠️ Search failed: {e}")

    state["retrieved_docs"] = retrieved_docs

    print(f"\n📚 Total documents retrieved: {len(retrieved_docs)}")

    return state


print("✅ Retrieval Agent created")
````

````python
from ddgs import DDGS

def retrieval_agent(state: ResearchState) -> ResearchState:
    """
    Search the web for each research sub-question
    and collect the retrieved documents.
    """

    retrieved_docs = []

    research_plan = state.get("research_plan", [])

    for sub_question in research_plan[:4]:

        print(f"\n🔍 Searching: {sub_question}")

        try:
            with DDGS() as ddgs:
                results = list(
                    ddgs.text(
                        sub_question,
                        max_results=3
                    )
                )

            for result in results:
                document = {
                    "title": result.get("title", "Unknown"),
                    "snippet": result.get("body", ""),
                    "url": result.get("href", ""),
                    "query": sub_question,
                    "retrieved_at": datetime.now().isoformat()
                }

                retrieved_docs.append(document)

        except Exception as e:
            print(f"⚠️ Search failed: {e}")

    state["retrieved_docs"] = retrieved_docs

    print(f"\n📚 Total documents retrieved: {len(retrieved_docs)}")

    return state


print("✅ Updated Retrieval Agent created")
````

````python
def reasoning_agent(state: ResearchState) -> ResearchState:
    """
    Synthesize retrieved research evidence into a final answer.
    """

    query = state["query"]
    retrieved_docs = state.get("retrieved_docs", [])

    if not retrieved_docs:
        state["final_answer"] = (
            "I could not retrieve sufficient sources to answer this question."
        )
        state["citations"] = []
        return state

    # Build evidence context
    context_parts = []

    for i, doc in enumerate(retrieved_docs[:10], 1):
        context_parts.append(
            f"""
SOURCE {i}
Title: {doc.get("title", "Unknown")}
URL: {doc.get("url", "")}
Research Question: {doc.get("query", "")}
Content: {doc.get("snippet", "")}
"""
        )

    context = "\n".join(context_parts)

    prompt = f"""
You are the Reasoning Agent in ResearchFlow AI, a multi-agent
research assistant.

USER QUERY:
{query}

RETRIEVED RESEARCH SOURCES:
{context}

TASK:
Synthesize the retrieved information into a clear, evidence-based
research answer.

REQUIREMENTS:

1. Answer the user's original query directly.
2. Organize the answer using clear headings.
3. Compare important approaches, findings, or techniques where
   appropriate.
4. Use ONLY information supported by the retrieved sources.
5. Do not invent facts, statistics, papers, datasets, or results.
6. If the retrieved information is insufficient for a claim,
   explicitly say that the available evidence is insufficient.
7. When mentioning a source, refer to it using [Source N], where
   N corresponds to the source number above.
8. End with a short "Key Takeaways" section.
9. Do not include a separate bibliography; citations will be
   generated by the system.
10. Do not use Markdown code fences.

Return only the final research answer.
"""

    response = llm.invoke([HumanMessage(content=prompt)])

    state["reasoning"] = response.content
    state["final_answer"] = response.content

    # Store citation metadata separately
    state["citations"] = [
        {
            "source_id": str(i),
            "title": doc.get("title", "Unknown"),
            "url": doc.get("url", "")
        }
        for i, doc in enumerate(retrieved_docs[:10], 1)
    ]

    return state


print("✅ Reasoning Agent created")
````

````python
reasoning_state: ResearchState = {
    "query": retrieval_state["query"],
    "research_plan": retrieval_state["research_plan"],
    "retrieved_docs": retrieval_state["retrieved_docs"]
}

reasoning_state = reasoning_agent(reasoning_state)

print("\n" + "=" * 80)
print("🧠 RESEARCH ANSWER")
print("=" * 80)

print(reasoning_state["final_answer"])

print("\n" + "=" * 80)
print("📚 CITATIONS")
print("=" * 80)

for citation in reasoning_state["citations"]:
    print(
        f"[Source {citation['source_id']}] "
        f"{citation['title']}"
    )
    print(f"URL: {citation['url']}\n")

print("✅ Reasoning test completed")
````

````python
def build_research_graph():
    """
    Build the ResearchFlow AI LangGraph workflow.
    """

    workflow = StateGraph(ResearchState)

    # Add agents
    workflow.add_node("query_planner", query_planner_agent)
    workflow.add_node("retrieval", retrieval_agent)
    workflow.add_node("reasoning", reasoning_agent)

    # Define workflow connections
    workflow.add_edge("query_planner", "retrieval")
    workflow.add_edge("retrieval", "reasoning")
    workflow.add_edge("reasoning", END)

    # Define starting point
    workflow.set_entry_point("query_planner")

    # Compile graph
    return workflow.compile()


research_graph = build_research_graph()

print("✅ ResearchFlow LangGraph compiled successfully")
print("\nWorkflow:")
print("🧠 Query Planner → 🔎 Retrieval → 🧠 Reasoning → END")
````

````python
import sqlite3
from datetime import datetime


class ResearchDatabase:
    """
    SQLite database for storing ResearchFlow queries,
    retrieved documents, and agent memory.
    """

    def __init__(self, db_path="researchflow.db"):
        self.db_path = db_path
        self._initialize_database()

    def _get_connection(self):
        return sqlite3.connect(self.db_path)

    def _initialize_database(self):
        conn = self._get_connection()
        cursor = conn.cursor()

        # Store research queries and final answers
        cursor.execute("""
            CREATE TABLE IF NOT EXISTS queries (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                query TEXT NOT NULL,
                answer TEXT,
                created_at TEXT NOT NULL
            )
        """)

        # Store retrieved research documents
        cursor.execute("""
            CREATE TABLE IF NOT EXISTS documents (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                query_id INTEGER NOT NULL,
                title TEXT,
                snippet TEXT,
                url TEXT,
                research_question TEXT,
                retrieved_at TEXT,
                FOREIGN KEY (query_id) REFERENCES queries(id)
            )
        """)

        # Store reusable agent memory
        cursor.execute("""
            CREATE TABLE IF NOT EXISTS memory (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                memory_key TEXT UNIQUE NOT NULL,
                memory_value TEXT,
                created_at TEXT NOT NULL,
                updated_at TEXT NOT NULL
            )
        """)

        conn.commit()
        conn.close()


    def save_query(self, query, answer):
        """Save a research query and its final answer."""

        conn = self._get_connection()
        cursor = conn.cursor()

        cursor.execute("""
            INSERT INTO queries (query, answer, created_at)
            VALUES (?, ?, ?)
        """, (
            query,
            answer,
            datetime.now().isoformat()
        ))

        query_id = cursor.lastrowid

        conn.commit()
        conn.close()

        return query_id


    def save_document(self, query_id, document):
        """Save a retrieved research document."""

        conn = self._get_connection()
        cursor = conn.cursor()

        cursor.execute("""
            INSERT INTO documents (
                query_id,
                title,
                snippet,
                url,
                research_question,
                retrieved_at
            )
            VALUES (?, ?, ?, ?, ?, ?)
        """, (
            query_id,
            document.get("title", ""),
            document.get("snippet", ""),
            document.get("url", ""),
            document.get("query", ""),
            document.get("retrieved_at", datetime.now().isoformat())
        ))

        conn.commit()
        conn.close()


    def save_memory(self, key, value):
        """Create or update a memory entry."""

        conn = self._get_connection()
        cursor = conn.cursor()

        now = datetime.now().isoformat()

        cursor.execute("""
            INSERT INTO memory (
                memory_key,
                memory_value,
                created_at,
                updated_at
            )
            VALUES (?, ?, ?, ?)
            ON CONFLICT(memory_key)
            DO UPDATE SET
                memory_value = excluded.memory_value,
                updated_at = excluded.updated_at
        """, (
            key,
            value,
            now,
            now
        ))

        conn.commit()
        conn.close()


    def get_memory(self, key):
        """Retrieve a memory entry."""

        conn = self._get_connection()
        cursor = conn.cursor()

        cursor.execute("""
            SELECT memory_value
            FROM memory
            WHERE memory_key = ?
        """, (key,))

        result = cursor.fetchone()

        conn.close()

        return result[0] if result else None


    def get_query(self, query_id):
        """Retrieve a saved research query."""

        conn = self._get_connection()
        cursor = conn.cursor()

        cursor.execute("""
            SELECT id, query, answer, created_at
            FROM queries
            WHERE id = ?
        """, (query_id,))

        result = cursor.fetchone()

        conn.close()

        return result


    def get_document_count(self, query_id):
        """Count documents associated with a research query."""

        conn = self._get_connection()
        cursor = conn.cursor()

        cursor.execute("""
            SELECT COUNT(*)
            FROM documents
            WHERE query_id = ?
        """, (query_id,))

        count = cursor.fetchone()[0]

        conn.close()

        return count


# Initialize database
db = ResearchDatabase()

print("✅ SQLite database initialized")
print(f"📁 Database file: {db.db_path}")
print("📊 Tables created: queries, documents, memory")
````

````python
def memory_aware_query_planner_agent(
    state: ResearchState
) -> ResearchState:
    """
    Query Planner that can use previous research memory
    as contextual information.
    """

    query = state["query"]

    # Retrieve previous research topic from memory
    previous_topic = db.get_memory("last_research_topic")

    if previous_topic:
        memory_context = f"""
Previous research topic stored in memory:
{previous_topic}

Use this only as contextual information.
Do not assume the previous research answers are correct,
and do not let the previous topic override the current query.
"""
    else:
        memory_context = """
No previous research topic is available.
"""

    prompt = f"""
You are the Query Planning Agent in ResearchFlow AI.

CURRENT USER QUERY:
{query}

MEMORY CONTEXT:
{memory_context}

TASK:
Break the current query into 3 to 4 specific research
sub-questions.

Requirements:
- Focus primarily on the CURRENT USER QUERY.
- Use memory only when it is genuinely relevant.
- Each question should investigate a different aspect.
- Avoid duplicate questions.
- Make each question specific and researchable.
- Return ONLY valid JSON.
- Do not use Markdown or code fences.

Required format:

{{
    "sub_questions": [
        "Question 1",
        "Question 2",
        "Question 3",
        "Question 4"
    ]
}}
"""

    response = llm.invoke([HumanMessage(content=prompt)])
    content = response.content.strip()

    try:
        if content.startswith("```"):
            content = (
                content
                .replace("```json", "")
                .replace("```", "")
                .strip()
            )

        plan_obj = json.loads(content)

        sub_questions = plan_obj.get("sub_questions", [])

        if not isinstance(sub_questions, list) or not sub_questions:
            raise ValueError("Invalid sub_questions format")

        state["research_plan"] = sub_questions[:4]

    except Exception as e:
        print(f"⚠️ Planner JSON parsing issue: {e}")
        state["research_plan"] = [query]

    # Store the current query as the latest research topic
    db.save_memory("last_research_topic", query)

    return state


print("✅ Memory-aware Query Planner created")
````

````python
def build_memory_aware_research_graph():
    """
    Build the ResearchFlow AI workflow with memory-aware planning.
    """

    workflow = StateGraph(ResearchState)

    # Memory-aware planner
    workflow.add_node(
        "query_planner",
        memory_aware_query_planner_agent
    )

    # Web retrieval
    workflow.add_node(
        "retrieval",
        retrieval_agent
    )

    # Reasoning and synthesis
    workflow.add_node(
        "reasoning",
        reasoning_agent
    )

    # Workflow connections
    workflow.add_edge(
        "query_planner",
        "retrieval"
    )

    workflow.add_edge(
        "retrieval",
        "reasoning"
    )

    workflow.add_edge(
        "reasoning",
        END
    )

    # Entry point
    workflow.set_entry_point("query_planner")

    return workflow.compile()


# Compile the memory-aware graph
research_graph = build_memory_aware_research_graph()

print("✅ Memory-aware ResearchFlow graph compiled")
print()
print("Workflow:")
print("🧠 Memory Planner → 🔎 Retrieval → 🧠 Reasoning → END")
````

````python
def deduplicate_documents(documents: List[Dict[str, Any]]) -> List[Dict[str, Any]]:
    """
    Remove duplicate research documents using their URLs.
    Falls back to title when a URL is unavailable.
    """

    unique_documents = []
    seen = set()

    for document in documents:

        url = document.get("url", "").strip()
        title = document.get("title", "").strip().lower()

        # Prefer URL as the unique identifier
        identifier = url if url else title

        if not identifier:
            continue

        if identifier in seen:
            continue

        seen.add(identifier)
        unique_documents.append(document)

    return unique_documents


print("✅ Document deduplication function created")
````

````python
def retrieval_agent(state: ResearchState) -> ResearchState:
    """
    Search the web for each research sub-question,
    collect documents, and remove duplicate sources.
    """

    retrieved_docs = []

    research_plan = state.get("research_plan", [])

    for sub_question in research_plan[:4]:

        print(f"\n🔍 Searching: {sub_question}")

        try:
            with DDGS() as ddgs:
                results = list(
                    ddgs.text(
                        sub_question,
                        max_results=3
                    )
                )

            for result in results:
                document = {
                    "title": result.get("title", "Unknown"),
                    "snippet": result.get("body", ""),
                    "url": result.get("href", ""),
                    "query": sub_question,
                    "retrieved_at": datetime.now().isoformat()
                }

                retrieved_docs.append(document)

        except Exception as e:
            print(f"⚠️ Search failed: {e}")

    # Remove duplicate sources
    unique_docs = deduplicate_documents(retrieved_docs)

    state["retrieved_docs"] = unique_docs

    print("\n" + "-" * 70)
    print("📊 RETRIEVAL SUMMARY")
    print("-" * 70)
    print(f"Raw documents retrieved:    {len(retrieved_docs)}")
    print(f"Unique documents retained:  {len(unique_docs)}")
    print(f"Duplicates removed:         {len(retrieved_docs) - len(unique_docs)}")

    return state


print("✅ Retrieval Agent upgraded with automatic deduplication")
````

````python
def build_research_graph():
    """
    Build the current ResearchFlow AI LangGraph.

    Flow:
    Memory-aware Planner → Deduplicated Retrieval → Reasoning
    """

    workflow = StateGraph(ResearchState)

    workflow.add_node(
        "query_planner",
        memory_aware_query_planner_agent
    )

    workflow.add_node(
        "retrieval",
        retrieval_agent
    )

    workflow.add_node(
        "reasoning",
        reasoning_agent
    )

    workflow.add_edge(
        "query_planner",
        "retrieval"
    )

    workflow.add_edge(
        "retrieval",
        "reasoning"
    )

    workflow.add_edge(
        "reasoning",
        END
    )

    workflow.set_entry_point("query_planner")

    return workflow.compile()


research_graph = build_research_graph()

print("✅ Final ResearchFlow graph compiled")
print()
print("Pipeline:")
print("🧠 Memory Planner")
print("       ↓")
print("🔎 Retrieval + Deduplication")
print("       ↓")
print("🧠 Groq Reasoning")
print("       ↓")
print("📄 Answer + Citations")
````

````python
# ============================================================
# ResearchFlow AI — Restore Runtime Components
# ============================================================

import os
import json
import sqlite3

from datetime import datetime
from typing import Any, Dict, List
from typing_extensions import TypedDict

from ddgs import DDGS
from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage
from langgraph.graph import StateGraph, END


# ------------------------------------------------------------
# 1. Research State
# ------------------------------------------------------------

class ResearchState(TypedDict, total=False):
    query: str
    research_plan: List[str]
    search_results: Dict[str, Any]
    retrieved_docs: List[Dict[str, Any]]
    reasoning: str
    final_answer: str
    citations: List[Dict[str, str]]
    memory: Dict[str, Any]


# ------------------------------------------------------------
# 2. Groq LLM
# ------------------------------------------------------------

if "GROQ_API_KEY" not in os.environ:
    from google.colab import userdata
    os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0.3,
    api_key=os.environ["GROQ_API_KEY"]
)


# ------------------------------------------------------------
# 3. SQLite Database
# ------------------------------------------------------------

class ResearchDatabase:

    def __init__(self, db_path="researchflow.db"):
        self.db_path = db_path
        self._initialize_database()

    def _get_connection(self):
        return sqlite3.connect(self.db_path)

    def _initialize_database(self):
        conn = self._get_connection()
        cursor = conn.cursor()

        cursor.execute("""
            CREATE TABLE IF NOT EXISTS queries (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                query TEXT NOT NULL,
                answer TEXT,
                created_at TEXT NOT NULL
            )
        """)

        cursor.execute("""
            CREATE TABLE IF NOT EXISTS documents (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                query_id INTEGER NOT NULL,
                title TEXT,
                snippet TEXT,
                url TEXT,
                research_question TEXT,
                retrieved_at TEXT,
                FOREIGN KEY (query_id) REFERENCES queries(id)
            )
        """)

        cursor.execute("""
            CREATE TABLE IF NOT EXISTS memory (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                memory_key TEXT UNIQUE NOT NULL,
                memory_value TEXT,
                created_at TEXT NOT NULL,
                updated_at TEXT NOT NULL
            )
        """)

        conn.commit()
        conn.close()

    def save_query(self, query, answer):
        conn = self._get_connection()
        cursor = conn.cursor()

        cursor.execute("""
            INSERT INTO queries (query, answer, created_at)
            VALUES (?, ?, ?)
        """, (query, answer, datetime.now().isoformat()))

        query_id = cursor.lastrowid

        conn.commit()
        conn.close()

        return query_id

    def save_document(self, query_id, document):
        conn = self._get_connection()
        cursor = conn.cursor()

        cursor.execute("""
            INSERT INTO documents (
                query_id,
                title,
                snippet,
                url,
                research_question,
                retrieved_at
            )
            VALUES (?, ?, ?, ?, ?, ?)
        """, (
            query_id,
            document.get("title", ""),
            document.get("snippet", ""),
            document.get("url", ""),
            document.get("query", ""),
            document.get(
                "retrieved_at",
                datetime.now().isoformat()
            )
        ))

        conn.commit()
        conn.close()

    def save_memory(self, key, value):
        conn = self._get_connection()
        cursor = conn.cursor()

        now = datetime.now().isoformat()

        cursor.execute("""
            INSERT INTO memory (
                memory_key,
                memory_value,
                created_at,
                updated_at
            )
            VALUES (?, ?, ?, ?)
            ON CONFLICT(memory_key)
            DO UPDATE SET
                memory_value = excluded.memory_value,
                updated_at = excluded.updated_at
        """, (key, value, now, now))

        conn.commit()
        conn.close()

    def get_memory(self, key):
        conn = self._get_connection()
        cursor = conn.cursor()

        cursor.execute("""
            SELECT memory_value
            FROM memory
            WHERE memory_key = ?
        """, (key,))

        result = cursor.fetchone()

        conn.close()

        return result[0] if result else None

    def get_query(self, query_id):
        conn = self._get_connection()
        cursor = conn.cursor()

        cursor.execute("""
            SELECT id, query, answer, created_at
            FROM queries
            WHERE id = ?
        """, (query_id,))

        result = cursor.fetchone()

        conn.close()

        return result

    def get_document_count(self, query_id):
        conn = self._get_connection()
        cursor = conn.cursor()

        cursor.execute("""
            SELECT COUNT(*)
            FROM documents
            WHERE query_id = ?
        """, (query_id,))

        count = cursor.fetchone()[0]

        conn.close()

        return count


db = ResearchDatabase()


# ------------------------------------------------------------
# 4. Document Deduplication
# ------------------------------------------------------------

def deduplicate_documents(
    documents: List[Dict[str, Any]]
) -> List[Dict[str, Any]]:

    unique_documents = []
    seen = set()

    for document in documents:

        url = document.get("url", "").strip()
        title = document.get("title", "").strip().lower()

        identifier = url if url else title

        if not identifier:
            continue

        if identifier in seen:
            continue

        seen.add(identifier)
        unique_documents.append(document)

    return unique_documents


# ------------------------------------------------------------
# 5. Memory-Aware Query Planner
# ------------------------------------------------------------

def memory_aware_query_planner_agent(
    state: ResearchState
) -> ResearchState:

    query = state["query"]

    previous_topic = db.get_memory(
        "last_research_topic"
    )

    if previous_topic:
        memory_context = f"""
Previous research topic stored in memory:
{previous_topic}

Use this only as contextual information.
Do not assume the previous research is correct.
Do not let it override the current query.
"""
    else:
        memory_context = """
No previous research topic is available.
"""

    prompt = f"""
You are the Query Planning Agent in ResearchFlow AI.

CURRENT USER QUERY:
{query}

MEMORY CONTEXT:
{memory_context}

Break the current query into 3 to 4 specific
research sub-questions.

Requirements:
- Focus primarily on the current query.
- Use memory only when genuinely relevant.
- Each question must investigate a different aspect.
- Avoid duplicates.
- Make questions specific and researchable.
- Return ONLY valid JSON.
- Do not use Markdown or code fences.

Required format:

{{
    "sub_questions": [
        "Question 1",
        "Question 2",
        "Question 3",
        "Question 4"
    ]
}}
"""

    response = llm.invoke(
        [HumanMessage(content=prompt)]
    )

    content = response.content.strip()

    try:
        if content.startswith("```"):
            content = (
                content
                .replace("```json", "")
                .replace("```", "")
                .strip()
            )

        plan_obj = json.loads(content)

        sub_questions = plan_obj.get(
            "sub_questions",
            []
        )

        if not isinstance(sub_questions, list):
            raise ValueError("Invalid sub_questions")

        if not sub_questions:
            raise ValueError("Empty research plan")

        state["research_plan"] = sub_questions[:4]

    except Exception as e:
        print(f"⚠️ Planner parsing issue: {e}")
        state["research_plan"] = [query]

    db.save_memory(
        "last_research_topic",
        query
    )

    return state


# ------------------------------------------------------------
# 6. Retrieval Agent
# ------------------------------------------------------------

def retrieval_agent(
    state: ResearchState
) -> ResearchState:

    retrieved_docs = []

    research_plan = state.get(
        "research_plan",
        []
    )

    for sub_question in research_plan[:4]:

        print(f"\n🔍 Searching: {sub_question}")

        try:
            with DDGS() as ddgs:

                results = list(
                    ddgs.text(
                        sub_question,
                        max_results=3
                    )
                )

            for result in results:

                document = {
                    "title": result.get(
                        "title",
                        "Unknown"
                    ),
                    "snippet": result.get(
                        "body",
                        ""
                    ),
                    "url": result.get(
                        "href",
                        ""
                    ),
                    "query": sub_question,
                    "retrieved_at":
                        datetime.now().isoformat()
                }

                retrieved_docs.append(document)

        except Exception as e:
            print(f"⚠️ Search failed: {e}")

    unique_docs = deduplicate_documents(
        retrieved_docs
    )

    state["retrieved_docs"] = unique_docs

    print("\n" + "-" * 70)
    print("📊 RETRIEVAL SUMMARY")
    print("-" * 70)
    print(
        f"Raw documents retrieved: "
        f"{len(retrieved_docs)}"
    )
    print(
        f"Unique documents retained: "
        f"{len(unique_docs)}"
    )
    print(
        f"Duplicates removed: "
        f"{len(retrieved_docs) - len(unique_docs)}"
    )

    return state


# ------------------------------------------------------------
# 7. Reasoning Agent
# ------------------------------------------------------------

def reasoning_agent(
    state: ResearchState
) -> ResearchState:

    query = state["query"]

    retrieved_docs = state.get(
        "retrieved_docs",
        []
    )

    if not retrieved_docs:

        state["final_answer"] = (
            "I could not retrieve sufficient "
            "sources to answer this question."
        )

        state["citations"] = []

        return state

    context_parts = []

    for i, doc in enumerate(
        retrieved_docs[:10],
        1
    ):

        context_parts.append(
            f"""
SOURCE {i}
Title: {doc.get("title", "Unknown")}
URL: {doc.get("url", "")}
Research Question: {doc.get("query", "")}
Content: {doc.get("snippet", "")}
"""
        )

    context = "\n".join(context_parts)

    prompt = f"""
You are the Reasoning Agent in ResearchFlow AI.

USER QUERY:
{query}

RETRIEVED RESEARCH SOURCES:
{context}

Synthesize the retrieved information into a
clear, evidence-based research answer.

Requirements:
1. Answer the original query directly.
2. Use clear headings.
3. Compare approaches where appropriate.
4. Use ONLY information supported by the sources.
5. Do not invent facts or statistics.
6. If evidence is insufficient, say so.
7. Cite sources using [Source N].
8. End with a "Key Takeaways" section.
9. Do not include a separate bibliography.
10. Do not use Markdown code fences.

Return only the final research answer.
"""

    response = llm.invoke(
        [HumanMessage(content=prompt)]
    )

    state["reasoning"] = response.content
    state["final_answer"] = response.content

    state["citations"] = [
        {
            "source_id": str(i),
            "title": doc.get(
                "title",
                "Unknown"
            ),
            "url": doc.get(
                "url",
                ""
            )
        }
        for i, doc in enumerate(
            retrieved_docs[:10],
            1
        )
    ]

    return state


# ------------------------------------------------------------
# 8. Build Final LangGraph
# ------------------------------------------------------------

def build_research_graph():

    workflow = StateGraph(ResearchState)

    workflow.add_node(
        "query_planner",
        memory_aware_query_planner_agent
    )

    workflow.add_node(
        "retrieval",
        retrieval_agent
    )

    workflow.add_node(
        "reasoning",
        reasoning_agent
    )

    workflow.add_edge(
        "query_planner",
        "retrieval"
    )

    workflow.add_edge(
        "retrieval",
        "reasoning"
    )

    workflow.add_edge(
        "reasoning",
        END
    )

    workflow.set_entry_point(
        "query_planner"
    )

    return workflow.compile()


research_graph = build_research_graph()


# ------------------------------------------------------------
# Restoration Summary
# ------------------------------------------------------------

print("=" * 70)
print("✅ RESEARCHFLOW RUNTIME RESTORED")
print("=" * 70)

print("✅ ResearchState")
print("✅ Groq LLM")
print("✅ SQLite Database")
print("✅ Memory-aware Query Planner")
print("✅ DDGS Retrieval")
print("✅ Document Deduplication")
print("✅ Reasoning Agent")
print("✅ LangGraph")

print("\n🚀 ResearchFlow is ready for testing.")
````
